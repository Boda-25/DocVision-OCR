````python
import pandas as pd

annotation_path = "/kaggle/input/datasets/nibinv23/iam-handwriting-word-database/iam_words/words.txt"

data = []

with open(annotation_path, "r", encoding="utf-8") as file:
    for line in file:
        line = line.strip()

        # Skip comments and empty lines
        if not line or line.startswith("#"):
            continue

        parts = line.split()

        # Keep the actual annotation fields
        if len(parts) >= 9:
            data.append(parts[:9])

# Create a clean DataFrame
df_annotations = pd.DataFrame(
    data,
    columns=[
        "word_id",
        "status",
        "graylevel",
        "x",
        "y",
        "width",
        "height",
        "tag",
        "transcription"
    ]
)

print("Number of annotation records:", len(df_annotations))
print("Number of columns:", df_annotations.shape[1])

print("\nData types:")
print(df_annotations.dtypes)

print("\nFirst 5 records:")
display(df_annotations.head())
````

````python
import os

images_root = "/kaggle/input/datasets/nibinv23/iam-handwriting-word-database/iam_words/words"

# Build image paths from word IDs
def get_image_path(word_id):
    parts = word_id.split("-")

    # Example:
    # a01-000u-00-00
    # -> a01 / a01-000u / a01-000u-00-00.png
    writer = parts[0]
    form = f"{parts[0]}-{parts[1]}"

    return os.path.join(
        images_root,
        writer,
        form,
        f"{word_id}.png"
    )

# Test the mapping
df_annotations["image_path"] = df_annotations["word_id"].apply(get_image_path)

# Check whether images exist
df_annotations["image_exists"] = df_annotations["image_path"].apply(os.path.exists)

print("Total annotations:", len(df_annotations))
print("Images found:", df_annotations["image_exists"].sum())
print("Images missing:", (~df_annotations["image_exists"]).sum())

print("\nSample mappings:")
display(
    df_annotations[
        ["word_id", "transcription", "image_path", "image_exists"]
    ].head(10)
)
````

## 3.4 — Image Dimensions Analysis

The physical dimensions of handwritten word images can vary significantly across samples.

Understanding the image width and height distribution is important because the OCR preprocessing pipeline will need to normalize images to a suitable input size while preserving the handwriting structure.

We will analyze:

- Image width
- Image height
- Aspect ratio
- Minimum and maximum dimensions
- Average image dimensions

````python
from PIL import Image, UnidentifiedImageError

widths = []
heights = []
valid_paths = []
corrupted_paths = []

for image_path in df_annotations["image_path"]:

    try:
        with Image.open(image_path) as img:
            width, height = img.size

        widths.append(width)
        heights.append(height)
        valid_paths.append(image_path)

    except (UnidentifiedImageError, OSError) as e:
        corrupted_paths.append(image_path)

# Create a mask for valid images
valid_mask = df_annotations["image_path"].isin(valid_paths)

# Keep only valid image records for dimension analysis
df_dimensions = df_annotations[valid_mask].copy()

# Store dimensions
df_dimensions["image_width"] = widths
df_dimensions["image_height"] = heights

# Calculate aspect ratio
df_dimensions["aspect_ratio"] = (
    df_dimensions["image_width"] /
    df_dimensions["image_height"]
)

print("===== IMAGE DIMENSIONS ANALYSIS =====")

print(f"Total annotation records: {len(df_annotations):,}")
print(f"Valid images: {len(df_dimensions):,}")
print(f"Corrupted/unreadable images: {len(corrupted_paths):,}")

print()

print("===== WIDTH =====")
print(f"Minimum width: {df_dimensions['image_width'].min()} px")
print(f"Maximum width: {df_dimensions['image_width'].max()} px")
print(f"Average width: {df_dimensions['image_width'].mean():.2f} px")

print()

print("===== HEIGHT =====")
print(f"Minimum height: {df_dimensions['image_height'].min()} px")
print(f"Maximum height: {df_dimensions['image_height'].max()} px")
print(f"Average height: {df_dimensions['image_height'].mean():.2f} px")

print()

print("===== ASPECT RATIO =====")
print(f"Minimum aspect ratio: {df_dimensions['aspect_ratio'].min():.2f}")
print(f"Maximum aspect ratio: {df_dimensions['aspect_ratio'].max():.2f}")
print(f"Average aspect ratio: {df_dimensions['aspect_ratio'].mean():.2f}")

print()

print("===== CORRUPTED FILES =====")
for path in corrupted_paths[:10]:
    print(path)
````

## 4.2 — Image Resizing Strategy

Handwritten word images have different widths and heights.

Deep Learning models require inputs with a consistent spatial size, so the images need to be resized before training.

However, directly stretching every image to the same width and height can distort the handwriting.

Therefore, our preprocessing strategy will:

- Normalize the image height to a fixed value
- Preserve the original aspect ratio
- Add padding to reach the target width
- Avoid unnecessary distortion of handwritten characters

The target image size will be:

**Height = 64 pixels**

The width will be normalized to a fixed maximum size after preserving the original aspect ratio.

````python
# Image resizing configuration

TARGET_HEIGHT = 64
TARGET_WIDTH = 256

print("===== IMAGE RESIZING CONFIGURATION =====")
print(f"Target height: {TARGET_HEIGHT} px")
print(f"Target width: {TARGET_WIDTH} px")
print("Aspect ratio preservation: Enabled")
print("Padding: Enabled")
````

## 4.3 — Image Preprocessing Function

We will now implement the core image preprocessing function.

For each handwritten word image, the function will:

1. Load the image in grayscale
2. Resize it to a fixed height of 64 pixels
3. Preserve the original aspect ratio
4. Add padding to reach a width of 256 pixels
5. Normalize pixel values to the range [0, 1]
6. Return the processed image as a NumPy array

This function will later be used to prepare images for the OCR deep learning model.

````python
import cv2
import numpy as np

def preprocess_image(image_path, target_height=64, target_width=256):
    """
    Preprocess a handwritten word image for OCR.

    Steps:
    - Load grayscale image
    - Resize while preserving aspect ratio
    - Pad to target dimensions
    - Normalize pixel values to [0, 1]
    """

    # Load image as grayscale
    image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

    if image is None:
        raise ValueError(f"Unable to read image: {image_path}")

    original_height, original_width = image.shape

    # Calculate new width while preserving aspect ratio
    new_width = int(
        original_width * (target_height / original_height)
    )

    # Resize image
    resized = cv2.resize(
        image,
        (new_width, target_height),
        interpolation=cv2.INTER_AREA
    )

    # If resized width is larger than target width,
    # resize it again to fit within the target width
    if new_width > target_width:

        resized = cv2.resize(
            image,
            (target_width, target_height),
            interpolation=cv2.INTER_AREA
        )

        new_width = target_width

    # Create white canvas
    canvas = np.ones(
        (target_height, target_width),
        dtype=np.uint8
    ) * 255

    # Center the resized image horizontally
    x_offset = (target_width - new_width) // 2

    canvas[
        :,
        x_offset:x_offset + new_width
    ] = resized

    # Normalize pixels to [0, 1]
    processed = canvas.astype(np.float32) / 255.0

    return processed
````

## 4.7 — Character Vocabulary Encoding

The OCR model cannot directly process text labels such as `"hello"` or `"computer"`.

We will create:

````python
# Build the character vocabulary

from collections import Counter

# Collect all characters from the clean dataset
all_text = "".join(
    df_clean["transcription"].astype(str)
)

# Extract unique characters
characters = sorted(set(all_text))

# Character -> integer
char_to_idx = {
    char: idx
    for idx, char in enumerate(characters)
}

# Integer -> character
idx_to_char = {
    idx: char
    for char, idx in char_to_idx.items()
}

print("===== CHARACTER VOCABULARY =====")

print(f"Number of characters: {len(characters)}")

print("\nCharacter to Index:")
print(char_to_idx)

print("\nVocabulary size:", len(char_to_idx))
````

## 4.9 — Text Label Encoding

The ground-truth transcriptions are currently stored as text strings.

For model training, each character must be converted into its corresponding integer ID using the character vocabulary created in the previous step.

For example:

`gets`

will be converted into:

`[56, 54, 69, 68]`

The CTC blank token will not be included in the ground-truth labels because it is handled internally by the CTC loss function.

````python
# Encode text transcriptions into integer sequences

def encode_text(text):
    """
    Convert a transcription into a sequence of character IDs.
    """
    return [
        char_to_idx[char]
        for char in text
    ]


# Test encoding
test_text = "gets"
encoded_text = encode_text(test_text)

print("===== TEXT ENCODING TEST =====")

print(f"Original text: {test_text}")
print(f"Encoded sequence: {encoded_text}")
print(f"Sequence length: {len(encoded_text)}")


# Apply encoding to the clean dataset
df_clean["label_encoded"] = df_clean["transcription"].apply(
    encode_text
)

print("\n===== DATASET LABEL ENCODING =====")
print(f"Encoded samples: {len(df_clean):,}")

display(
    df_clean[
        ["transcription", "label_encoded"]
    ].head(10)
)
````

## 5.3 — OCR Evaluation Metrics

To evaluate OCR predictions, we will implement two standard metrics:

### Character Error Rate (CER)

CER measures the number of character-level edits required to transform the predicted text into the ground-truth text.

It is calculated as:

CER = (Substitutions + Insertions + Deletions) / Number of Ground-Truth Characters

A lower CER indicates better character recognition.

### Word Error Rate (WER)

WER measures errors at the word level.

For our current word-level dataset, each image contains one handwritten word, so:

- Correct prediction → WER = 0
- Incorrect prediction → WER = 1

These metrics will be used to quantitatively evaluate the Tesseract baseline and later compare it with the Deep Learning OCR model.

````python
# OCR evaluation metrics

def edit_distance(reference, hypothesis):
    """
    Calculate Levenshtein edit distance
    between reference and hypothesis.
    """

    reference = str(reference)
    hypothesis = str(hypothesis)

    rows = len(reference) + 1
    cols = len(hypothesis) + 1

    dp = np.zeros(
        (rows, cols),
        dtype=np.int32
    )

    for i in range(rows):
        dp[i, 0] = i

    for j in range(cols):
        dp[0, j] = j

    for i in range(1, rows):
        for j in range(1, cols):

            if reference[i - 1] == hypothesis[j - 1]:
                cost = 0
            else:
                cost = 1

            dp[i, j] = min(
                dp[i - 1, j] + 1,       # Deletion
                dp[i, j - 1] + 1,       # Insertion
                dp[i - 1, j - 1] + cost # Substitution
            )

    return int(dp[-1, -1])


def character_error_rate(reference, hypothesis):
    """
    Calculate Character Error Rate (CER).
    """

    reference = str(reference)
    hypothesis = str(hypothesis)

    if len(reference) == 0:
        return 0.0 if len(hypothesis) == 0 else 1.0

    distance = edit_distance(
        reference,
        hypothesis
    )

    return distance / len(reference)


def word_error_rate(reference, hypothesis):
    """
    Calculate Word Error Rate (WER).
    """

    reference = str(reference).strip()
    hypothesis = str(hypothesis).strip()

    return 0.0 if reference == hypothesis else 1.0


# Test the metrics on our previous example

reference = "each"
hypothesis = "cack"

cer = character_error_rate(
    reference,
    hypothesis
)

wer = word_error_rate(
    reference,
    hypothesis
)

print("===== METRIC TEST =====")
print(f"Reference  : {reference}")
print(f"Hypothesis : {hypothesis}")
print(f"CER        : {cer:.4f}")
print(f"WER        : {wer:.4f}")
````

````python
# Create PyTorch Dataset and DataLoader

import torch
from torch.utils.data import Dataset, DataLoader

class OCRDataset(Dataset):
    def __init__(self, dataframe):
        self.dataframe = dataframe.reset_index(drop=True)

    def __len__(self):
        return len(self.dataframe)

    def __getitem__(self, index):
        row = self.dataframe.iloc[index]

        image = preprocess_image(
            row["image_path"],
            target_height=TARGET_HEIGHT,
            target_width=TARGET_WIDTH
        )

        image = torch.tensor(
            image,
            dtype=torch.float32
        ).unsqueeze(0)

        label = torch.tensor(
            row["label_encoded"],
            dtype=torch.long
        )

        return image, label


def ocr_collate_fn(batch):
    images, labels = zip(*batch)

    images = torch.stack(images, dim=0)

    label_lengths = torch.tensor(
        [len(label) for label in labels],
        dtype=torch.long
    )

    labels = torch.cat(labels)

    return images, labels, label_lengths


# Create datasets
train_dataset = OCRDataset(df_train)
val_dataset = OCRDataset(df_val)
test_dataset = OCRDataset(df_test)

# Small batch size because the current environment is CPU-only
BATCH_SIZE = 8

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    collate_fn=ocr_collate_fn,
    num_workers=0
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    collate_fn=ocr_collate_fn,
    num_workers=0
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    collate_fn=ocr_collate_fn,
    num_workers=0
)

print("===== DATASET & DATALOADER =====")

print(f"Train samples : {len(train_dataset):,}")
print(f"Val samples   : {len(val_dataset):,}")
print(f"Test samples  : {len(test_dataset):,}")

print(f"\nBatch size    : {BATCH_SIZE}")
print(f"Train batches : {len(train_loader):,}")
print(f"Val batches   : {len(val_loader):,}")
print(f"Test batches  : {len(test_loader):,}")
````

## 6.5 — CRNN Model Architecture

We will build a CRNN (Convolutional Recurrent Neural Network) for handwriting recognition.

### Architecture

Input Image
→ CNN Feature Extractor
→ Sequence Features
→ Bidirectional LSTM
→ Character Classification

The CNN extracts visual features from the handwritten word.

The Bidirectional LSTM processes the extracted features as a character sequence.

The final layer predicts one of the 77 CTC classes at each sequence timestep:

- 76 character classes
- 1 CTC blank class

CTC decoding will be handled during training and inference.

````python
# Build the CRNN OCR model

import torch
import torch.nn as nn


class CRNN(nn.Module):

    def __init__(
        self,
        num_classes,
        hidden_size=128
    ):
        super(CRNN, self).__init__()

        # CNN feature extractor
        self.cnn = nn.Sequential(

            # Block 1
            nn.Conv2d(
                1, 64,
                kernel_size=3,
                padding=1
            ),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(
                kernel_size=2,
                stride=2
            ),

            # Block 2
            nn.Conv2d(
                64, 128,
                kernel_size=3,
                padding=1
            ),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(
                kernel_size=2,
                stride=2
            ),

            # Block 3
            nn.Conv2d(
                128, 256,
                kernel_size=3,
                padding=1
            ),
            nn.ReLU(inplace=True),

            # Block 4
            nn.Conv2d(
                256, 256,
                kernel_size=3,
                padding=1
            ),
            nn.ReLU(inplace=True),

            nn.MaxPool2d(
                kernel_size=(2, 1),
                stride=(2, 1)
            )
        )

        # After CNN:
        # Height: 64 -> 32 -> 16 -> 8
        # Width : 256 -> 128 -> 64
        #
        # Feature map:
        # [Batch, 256, 8, 64]

        self.lstm = nn.LSTM(
            input_size=256 * 8,
            hidden_size=hidden_size,
            num_layers=2,
            bidirectional=True,
            batch_first=False
        )

        # Bidirectional LSTM => hidden_size * 2
        self.classifier = nn.Linear(
            hidden_size * 2,
            num_classes
        )

    def forward(self, x):

        # CNN
        features = self.cnn(x)

        # [B, C, H, W]
        batch_size, channels, height, width = features.size()

        # Convert width into sequence dimension
        # [B, C, H, W]
        # -> [W, B, C*H]
        features = features.permute(
            3, 0, 1, 2
        )

        features = features.reshape(
            width,
            batch_size,
            channels * height
        )

        # BiLSTM
        sequence, _ = self.lstm(features)

        # Character classification
        output = self.classifier(sequence)

        return output


# Create model
NUM_CLASSES = CTC_BLANK + 1

model = CRNN(
    num_classes=NUM_CLASSES,
    hidden_size=128
)

print("===== CRNN MODEL =====")
print(model)

print("\nTotal parameters:")

total_parameters = sum(
    parameter.numel()
    for parameter in model.parameters()
)

print(f"{total_parameters:,}")
````

## 7.1 — Training & Validation Pipeline

We will now implement the training pipeline for the CRNN OCR model.

The pipeline will:

- Train the CRNN using CTC Loss
- Calculate the average training loss
- Evaluate validation loss after each epoch
- Track training and validation history
- Keep the model in the correct training/evaluation mode

We will first validate the pipeline with a single epoch before running a longer training process.

````python
# Training and validation functions

def train_one_epoch(model, loader, optimizer, criterion, device):
    model.train()

    total_loss = 0.0
    total_samples = 0

    for images, labels, label_lengths in loader:

        images = images.to(device)
        labels = labels.to(device)
        label_lengths = label_lengths.to(device)

        optimizer.zero_grad()

        outputs = model(images)

        log_probs = outputs.log_softmax(dim=2)

        input_lengths = torch.full(
            size=(images.size(0),),
            fill_value=outputs.size(0),
            dtype=torch.long,
            device=device
        )

        loss = criterion(
            log_probs,
            labels,
            input_lengths,
            label_lengths
        )

        loss.backward()

        optimizer.step()

        batch_size = images.size(0)

        total_loss += loss.item() * batch_size
        total_samples += batch_size

    return total_loss / total_samples


def validate_one_epoch(model, loader, criterion, device):
    model.eval()

    total_loss = 0.0
    total_samples = 0

    with torch.no_grad():

        for images, labels, label_lengths in loader:

            images = images.to(device)
            labels = labels.to(device)
            label_lengths = label_lengths.to(device)

            outputs = model(images)

            log_probs = outputs.log_softmax(dim=2)

            input_lengths = torch.full(
                size=(images.size(0),),
                fill_value=outputs.size(0),
                dtype=torch.long,
                device=device
            )

            loss = criterion(
                log_probs,
                labels,
                input_lengths,
                label_lengths
            )

            batch_size = images.size(0)

            total_loss += loss.item() * batch_size
            total_samples += batch_size

    return total_loss / total_samples


print("===== TRAINING PIPELINE =====")
print("Training function  : Ready")
print("Validation function: Ready")
print("Optimizer          : Adam")
print("Loss               : CTC Loss")
print("Device             :", device)

print("\nTraining pipeline is ready.")
````

## 7.6 — Post-Training Prediction Check

Before continuing training, we evaluate the current CRNN model
qualitatively by decoding its CTC predictions on validation samples.

This helps verify whether the model has started learning meaningful
character patterns after Epoch 1.

````python
import torch
import numpy as np

idx_to_char = {
    index: char
    for char, index in char_to_idx.items()
}

def ctc_greedy_decode(output):
    """
    Decode CTC output using greedy decoding:
    1. Select the most probable class at each timestep.
    2. Remove consecutive duplicate predictions.
    3. Remove the CTC blank token.
    """
    predictions = output.argmax(dim=2)

    decoded_texts = []

    for sample in predictions.permute(1, 0):
        previous = None
        decoded = []

        for index in sample.cpu().numpy():
            index = int(index)

            if index == CTC_BLANK:
                previous = index
                continue

            if index != previous:
                decoded.append(idx_to_char[index])

            previous = index

        decoded_texts.append("".join(decoded))

    return decoded_texts


model.eval()

images, labels, label_lengths = next(iter(val_loader))

images = images.to(device)

with torch.no_grad():
    outputs = model(images)

predictions = ctc_greedy_decode(outputs)

# Decode ground-truth labels
ground_truths = []

offset = 0

for length in label_lengths.tolist():
    label = labels[offset:offset + length]
    text = "".join(
        idx_to_char[int(index)]
        for index in label
    )
    ground_truths.append(text)
    offset += length

print("===== CRNN EPOCH 1 PREDICTIONS =====\n")

for i in range(min(8, len(predictions))):
    print(f"Sample {i + 1}")
    print(f"Ground Truth : {ground_truths[i]}")
    print(f"Prediction   : {predictions[i]}")
    print("-" * 40)

print("\nPrediction check completed.")
````

## 7.14 — CTC Blank Collapse Recovery Experiment

The CRNN is currently dominated by the CTC blank class.

- Apply gradient clipping
- Train for only 200 updates

````python
import torch
import torch.nn as nn


class CRNN_BN(nn.Module):
    def __init__(self, num_classes, hidden_size=128):
        super(CRNN_BN, self).__init__()

        self.cnn = nn.Sequential(
            # Block 1
            nn.Conv2d(1, 64, kernel_size=3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2, 2),

            # Block 2
            nn.Conv2d(64, 128, kernel_size=3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2, 2),

            # Block 3
            nn.Conv2d(128, 256, kernel_size=3, padding=1),
            nn.BatchNorm2d(256),
            nn.ReLU(inplace=True),

            # Block 4
            nn.Conv2d(256, 256, kernel_size=3, padding=1),
            nn.BatchNorm2d(256),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(kernel_size=(2, 1), stride=(2, 1))
        )

        self.lstm = nn.LSTM(
            input_size=256 * 8,
            hidden_size=hidden_size,
            num_layers=2,
            bidirectional=True,
            batch_first=False
        )

        self.classifier = nn.Linear(
            hidden_size * 2,
            num_classes
        )

    def forward(self, x):
        features = self.cnn(x)

        batch_size, channels, height, width = features.size()

        features = features.permute(3, 0, 1, 2)

        features = features.reshape(
            width,
            batch_size,
            channels * height
        )

        sequence, _ = self.lstm(features)

        output = self.classifier(sequence)

        return output


# Create fresh model
bn_model = CRNN_BN(
    num_classes=NUM_CLASSES,
    hidden_size=128
).to(device)


# Parameter count
total_params = sum(
    parameter.numel()
    for parameter in bn_model.parameters()
)

print("===== CRNN + BATCH NORMALIZATION =====")
print(f"Total parameters: {total_params:,}")


# Forward pass
bn_model.eval()

images, labels, label_lengths = next(iter(val_ok_loader))
images = images.to(device)
labels = labels.to(device)
label_lengths = label_lengths.to(device)

with torch.no_grad():
    outputs = bn_model(images)


print("\n===== FORWARD PASS =====")
print(f"Input shape:  {images.shape}")
print(f"Output shape: {outputs.shape}")


# CTC loss
log_probs = outputs.log_softmax(dim=2)

input_lengths = torch.full(
    size=(images.size(0),),
    fill_value=outputs.size(0),
    dtype=torch.long,
    device=device
)

with torch.no_grad():
    initial_ctc_loss = ctc_loss(
        log_probs,
        labels,
        input_lengths,
        label_lengths
    )


print("\n===== INITIAL CTC LOSS =====")
print(f"CTC Loss: {initial_ctc_loss.item():.6f}")


# Initial output distribution
probabilities = outputs.softmax(dim=2)

mean_class_prob = probabilities.mean(dim=(0, 1))

top_values, top_indices = torch.topk(
    mean_class_prob,
    k=10
)

print("\n===== INITIAL OUTPUT DISTRIBUTION =====")

for rank, (prob, index) in enumerate(
    zip(
        top_values.cpu().tolist(),
        top_indices.cpu().tolist()
    ),
    start=1
):
    if index == CTC_BLANK:
        label = "<BLANK>"
    else:
        label = repr(idx_to_char[index])

    print(
        f"{rank:2d}. "
        f"Class {index:2d} "
        f"{label:10s} "
        f"Mean Probability: {prob:.6f}"
    )


blank_probability = mean_class_prob[CTC_BLANK].item()

print("\n===== INITIAL BLANK PROBABILITY =====")
print(f"CTC Blank mean probability: {blank_probability:.6f}")


print("\n===== TASK 7.22 COMPLETED =====")
````

````python
# ===========================================
# PHASE 7.26 — SINGLE BATCH CTC MEMORIZATION
# ===========================================

import torch
import torch.nn as nn


# ----------------------
# 1. Get ONE fixed batch
# ----------------------

fixed_images, fixed_labels, fixed_label_lengths = next(
    iter(short_loader)
)

fixed_images = fixed_images.to(device)
fixed_labels = fixed_labels.to(device)
fixed_label_lengths = fixed_label_lengths.to(device)

print("===== FIXED BATCH =====")
print(f"Images: {fixed_images.shape}")
print(f"Labels: {fixed_labels.shape}")
print(f"Label lengths: {fixed_label_lengths.tolist()}")


# -----------------------------
# 2. Decode ground-truth labels
# -----------------------------

ground_truth_texts = []

offset = 0

for length in fixed_label_lengths.cpu().tolist():

    encoded = fixed_labels[offset:offset + length]

    text = "".join(
        idx_to_char[int(index)]
        for index in encoded.cpu().tolist()
    )

    ground_truth_texts.append(text)

    offset += length


print("\nGround Truth:")

for i, text in enumerate(ground_truth_texts, start=1):
    print(f"{i}. {text!r}")


# --------------
# 3. Fresh CRNN
# --------------

single_batch_model = CRNN(
    num_classes=NUM_CLASSES,
    hidden_size=128
).to(device)


single_batch_optimizer = torch.optim.Adam(
    single_batch_model.parameters(),
    lr=1e-3
)


single_batch_ctc = nn.CTCLoss(
    blank=CTC_BLANK,
    zero_infinity=True
)


# -----------------------
# 4. Memorize fixed batch
# -----------------------

single_batch_model.train()

NUM_UPDATES = 1000

print("\n===== MEMORIZATION TRAINING =====\n")

for step in range(1, NUM_UPDATES + 1):

    single_batch_optimizer.zero_grad()

    outputs = single_batch_model(fixed_images)

    log_probs = outputs.log_softmax(dim=2)

    input_lengths = torch.full(
        size=(fixed_images.size(0),),
        fill_value=outputs.size(0),
        dtype=torch.long,
        device=device
    )

    loss = single_batch_ctc(
        log_probs,
        fixed_labels,
        input_lengths,
        fixed_label_lengths
    )

    loss.backward()

    torch.nn.utils.clip_grad_norm_(
        single_batch_model.parameters(),
        max_norm=5.0
    )

    single_batch_optimizer.step()

    if step % 100 == 0:

        print(
            f"Step {step:4d} | "
            f"Loss: {loss.item():.6f}"
        )


# --------------------
# 5. Final predictions
# --------------------

single_batch_model.eval()

with torch.no_grad():

    final_outputs = single_batch_model(
        fixed_images
    )

final_predictions = ctc_greedy_decode(
    final_outputs
)


print("\n===== FINAL MEMORIZATION RESULTS =====\n")

for i, (gt, pred) in enumerate(
    zip(ground_truth_texts, final_predictions),
    start=1
):

    print(
        f"{i}. "
        f"GT: {gt!r:10s} "
        f"| Prediction: {pred!r}"
    )


# -------------------------
# 6. Final blank diagnostic
# -------------------------

final_probabilities = final_outputs.softmax(dim=2)

final_mean_probability = final_probabilities.mean(
    dim=(0, 1)
)

final_blank_probability = (
    final_mean_probability[CTC_BLANK].item()
)

print("\n===== FINAL BLANK DIAGNOSTIC =====")
print(
    f"Blank mean probability: "
    f"{final_blank_probability:.6f}"
)

print("\n===== TASK 7.26 COMPLETED =====")
````

Create a reusable inference function that accepts
an image path and returns the recognized text.

Pipeline:

Image => TrOCR Processor => TrOCR Model => Text

````python
from PIL import Image
import torch

def ocr_image(image_path):
    """
    Run OCR inference on a single image.

    Parameters
    ----------
    image_path : str
        Path to the input image.

    Returns
    -------
    str
        Recognized text.
    """

    image = Image.open(image_path).convert("RGB")

    pixel_values = processor(
        images=image,
        return_tensors="pt"
    ).pixel_values.to(device)

    with torch.no_grad():
        generated_ids = trocr_model.generate(
            pixel_values
        )

    text = processor.batch_decode(
        generated_ids,
        skip_special_tokens=True
    )[0].strip()

    return text

print("OCR inference function created successfully!")
````

````python
# ====================================
# DocVision OCR — Custom Image Upload
# ====================================

import io
import ipywidgets as widgets
from IPython.display import display, clear_output
from PIL import Image
import matplotlib.pyplot as plt
import torch

# -------------
# Upload widget
# -------------

upload = widgets.FileUpload(
    accept="image/*",
    multiple=False,
    description=" Upload Image"
)

run_upload_button = widgets.Button(
    description=" Recognize Text",
    button_style="primary",
    layout=widgets.Layout(width="200px")
)

upload_output = widgets.Output()


# ------------
# OCR function
# ------------

def run_uploaded_ocr(button):

    with upload_output:

        clear_output(wait=True)

        if not upload.value:

            print(" Please upload an image first.")
            return

        # -----------------
        # Get uploaded file
        # -----------------

        uploaded_file = next(iter(upload.value))

        image_bytes = uploaded_file["content"]

        image = Image.open(
            io.BytesIO(image_bytes)
        ).convert("RGB")

        # ---------------
        # TrOCR inference
        # ---------------

        pixel_values = processor(
            images=image,
            return_tensors="pt"
        ).pixel_values.to(device)

        with torch.no_grad():

            generated_ids = trocr_model.generate(
                pixel_values
            )

        prediction = processor.batch_decode(
            generated_ids,
            skip_special_tokens=True
        )[0].strip()

        # -------------
        # Display image
        # -------------

        plt.figure(figsize=(12, 5))

        plt.imshow(image)
        plt.axis("off")

        plt.title(
            "Uploaded Handwriting Image",
            fontsize=15,
            fontweight="bold"
        )

        plt.show()

        # ------------------
        # Display prediction
        # ------------------

        display(
            widgets.HTML(
                f"""
                <div style="
                    padding:20px;
                    border:1px solid #ddd;
                    border-radius:12px;
                    margin-top:10px;
                ">

                <h2>TrOCR Prediction</h2>

                <p style="font-size:20px;">
                    <b>{prediction}</b>
                </p>

                </div>
                """
            )
        )


run_upload_button.on_click(
    run_uploaded_ocr
)


# ------------------------
# Display upload interface
# ------------------------

display(
    widgets.HTML(
        """
        <h2> Custom OCR Test</h2>

        <p>
        Upload a handwritten image and let
        <b>DocVision OCR</b> recognize the text.
        </p>
        """
    ),
    upload,
    run_upload_button,
    upload_output
)
````
